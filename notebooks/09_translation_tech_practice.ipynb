## 练习 1：平行句对与简单统计

读取 `examples/data/parallel_sample.tsv`，打印句对，并计算英文侧平均词数。


```python
from pathlib import Path

path = Path('../examples/data/parallel_sample.tsv')
lines = path.read_text(encoding='utf-8').splitlines()[1:]
pairs = [tuple(line.split('\t', maxsplit=1)) for line in lines if line.strip()]

word_counts = [len(en.split()) for en, _ in pairs]
avg_words = sum(word_counts) / len(word_counts)

for en, zh in pairs:
    print(f'[{len(en.split())} w] {en}')
    print(f'    {zh}\n')
print(f'英文平均词数: {avg_words:.1f}')
```

## 练习 2：术语表查找

用字典缓存 CSV 术语表，检查一句英文里是否包含表中的关键词。


```python
import csv
from pathlib import Path

terms_path = Path('../examples/data/terminology.csv')
terms: dict[str, str] = {}
with terms_path.open(encoding='utf-8', newline='') as f:
    for row in csv.DictReader(f):
        terms[row['source_en'].lower()] = row['target_zh']

sentence = 'A good post-editing workflow needs consistent terminology.'
tokens = {t.lower().strip('.,') for t in sentence.split()}
hits = {en: terms[en] for en in tokens if en in terms}
print('句子:', sentence)
print('命中术语:', hits)
```

## 练习 3：项目路径与文件遍历

模拟本地化项目目录，用 `pathlib.Path.rglob` 列出所有文本文件。


```python
from pathlib import Path

project = Path('../examples/project_layout')
for file in sorted(project.rglob('*.txt')):
    rel = file.relative_to(project)
    preview = file.read_text(encoding='utf-8').strip().replace('\n', ' / ')
    print(f'{rel}: {preview[:60]}')
```
