# 编程语言  

**编程语言**是一种语言，是一种计算机能读懂的语言。计算机理解不了表达不清（模糊，fuzzy）或模棱两可（歧义，ambiguous）的表述。比如，你去买个西瓜，你会自己决定去哪里买，买几个，找个搞活动打折的买。但是，你想让计算机买一个西瓜？你要告诉他：去门口的鲜丰水果店买，买沙瓤无子西瓜，若旁边的店打折，就去旁边的店买。总之，你不能让计算机做任何决定，你要清楚的告诉他所有情况下的所有的行为。而code，就是你和计算机交流的语言，或者说是对计算机的命令。

我们就把计算机理解为一个人，这个人很死板，只能按照你给他的详细命令进行工作。而这个人工作速度特别特别快，并且保证工作结果都是正确的。

编程语言其实不重要，重要的是要明白如何和计算机交流，明白了这个也就能看懂代码了。

编程 = 算法 + 数据结构
